# Project 02: Data Visualization with Matplotlib & Seaborn

**Dataset:** Seaborn Tips dataset (provided locally as `tips.csv`).

This notebook explores restaurant bills and tips using a histogram, box plot, scatter plot,
and annotated correlation heatmap. Each chart includes a finding-focused title, labelled axes,
and a short interpretation.

## 1. Import libraries and load the dataset

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

DATA_FILE = Path("tips.csv")
tips = pd.read_csv(DATA_FILE)

print("Dataset shape:", tips.shape)
display(tips.head())
display(tips.describe(include="all"))
```

## 2. Questions
- How are total bills distributed?
- Do tip amounts differ by day?
- Is total bill associated with tip amount?
- Which numerical variables move together?

```python
sns.set_theme(style="whitegrid", palette="deep")
fig, axes = plt.subplots(2, 2, figsize=(12, 9), constrained_layout=True)
fig.suptitle("Restaurant Tips Dataset — Data Visualization Dashboard",
             fontsize=16, fontweight="bold")

# Histogram: distribution of one numerical variable
sns.histplot(data=tips, x="total_bill", bins=20, color="#3569a8",
             edgecolor="white", ax=axes[0, 0])
axes[0, 0].set(title="Most bills are concentrated in the lower range",
               xlabel="Total bill (USD)", ylabel="Number of meals")

# Box plot: compare tip distributions across days
sns.boxplot(data=tips, x="day", y="tip", order=["Thur", "Fri", "Sat", "Sun"],
            color="#79a6d2", ax=axes[0, 1])
axes[0, 1].set(title="Tip amounts vary across days",
               xlabel="Day of week", ylabel="Tip (USD)")

# Scatter plot: relationship between two numerical variables
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="time",
                palette={"Lunch": "#2a9d8f", "Dinner": "#e76f51"},
                alpha=0.8, ax=axes[1, 0])
axes[1, 0].set(title="Larger bills generally come with larger tips",
               xlabel="Total bill (USD)", ylabel="Tip (USD)")
axes[1, 0].legend(title="Meal time")

# Correlation heatmap: numerical variables only
numeric_corr = tips.select_dtypes(include="number").corr()
sns.heatmap(numeric_corr, annot=True, fmt=".2f", cmap="Blues",
            vmin=-1, vmax=1, square=True, linewidths=0.5, ax=axes[1, 1])
axes[1, 1].set_title("Correlation between numerical variables")

plt.show()
```

## 3. Numerical findings and chart takeaways

```python
print(f"Number of transactions: {len(tips)}")
print(f"Median total bill: ${tips['total_bill'].median():.2f}")
print(f"Mean total bill: ${tips['total_bill'].mean():.2f}")
print(f"Total bill–tip correlation: {tips['total_bill'].corr(tips['tip']):.2f}")
print("\\nMedian tip by day:")
display(tips.groupby("day")["tip"].median().reindex(["Thur", "Fri", "Sat", "Sun"]))

print("Correlation matrix:")
display(numeric_corr.round(2))
```

### Takeaways

1. **Histogram:** Total bills are right-skewed, with most observations in the lower-to-middle range.
2. **Box plot:** Tip distributions overlap across days, but their medians vary; sample sizes should be considered.
3. **Scatter plot:** Higher total bills generally correspond to higher tips. Association does not prove causation.
4. **Heatmap:** The annotated matrix makes the strongest numerical relationships easy to identify; correlation measures linear association, not causation.
